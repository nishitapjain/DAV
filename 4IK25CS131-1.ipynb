```python
import numpy as np
```

```python
arr = np.array([10, 20, 30 , 40 ,50])

print("Numpy Array: ")
print(arr)
```

```python
print("\n--- Array Indexing ---")

print("First Element:", arr[0])
print("Third Element:", arr[2])
print("Last Element:", arr[-1])
```

```python
print("\n--- Array Slicing ---")

print("First Three Elements:", arr[0:3])
print("3rd and 4th Elements:",arr[2:4])
print("Element From Index 2:",arr[2:])
print("Last Three Elements:",arr[-3:])
```

```python
print("\n--- Basic Array Operation ---")

print("Add 5:", arr + 5)
print("Subtract 5:",arr-5)
print("Multiply by 2:",arr*2)
print("Divide by 10:", arr/10)
print("Divide by 2:",arr/2) #Division output is float
```

```python
print("\n--- Mathematical Operations ---")

print("Sum:",np.sum(arr))
print("Mean:",np.mean(arr))
print("Maximum:",np.max(arr))
print("Minimum:",np.min(arr))
```

```python
print("Original Array Shape") #original array shape is here
print(np.shape(arr))
print(arr)

print("\n--- Array Reshaping ---")
reshaped_arr = arr.reshape(5,1)

print("Reshaped Array:")
print(reshaped_arr)
```

```python
import pandas as pd
```

```python
data = {
    "Name" : ["Asha", "Ravi" , "Meera", "Keran"],
    "Marks" : [88, 72, 91, 65],
    "Branch" : ["CSE","ECE","CSE", "ISE"]
}
```

```python
df = pd.DataFrame(data)

print("\n--- Pandas DataFrame ---")
print(df)
```

```python
print("\n--- Column Indexing ---")

print("Name Column:")
print(df["Name"])

print("\nMarks Column:")
print(df["Marks"])
```

```python
print("\n--- Row Indexing ---")

print("First Row:")
print(df.iloc[0])

print("\nThird Row:")
print(df.iloc[2])
```

```python
print("\n--- Specific Value ---")

print("Markls Of First Student:",df.iloc[0,1])
```

```python
print("\n--- Row Slicing ---")

print("First Two Rows:")
print(df.iloc[0:2])
```

```python
print("\n--- Row and Column Slicing ---")

print("First Three Rows With Names and Marks:")
print(df.iloc[0:3,[0,1]])
```

```python
print("\n--- Data Filtering ---")

result = df[df["Marks"] > 70]

print("Student With Marks Greater than 70:")
print(result)
```
